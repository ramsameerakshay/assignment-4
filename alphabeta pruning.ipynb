```python
class Node:
    def __init__(self, value=None, children=None):
        self.value = value
        self.children = children or []


def alphabeta(node, depth, alpha, beta, maximizing):

    if depth == 0 or not node.children:
        return node.value

    if maximizing:
        value = float('-inf')

        for child in node.children:
            value = max(
                value,
                alphabeta(child, depth-1,
                          alpha, beta, False)
            )

            alpha = max(alpha, value)

            if beta <= alpha:
                break

        return value

    else:
        value = float('inf')

        for child in node.children:
            value = min(
                value,
                alphabeta(child, depth-1,
                          alpha, beta, True)
            )

            beta = min(beta, value)

            if beta <= alpha:
                break

        return value


root = Node(children=[
    Node(children=[Node(3), Node(5)]),
    Node(children=[Node(2), Node(9)])
])

result = alphabeta(root, 2,
                   float('-inf'),
                   float('inf'),
                   True)

print("Alpha Beta Result =", result)
```
